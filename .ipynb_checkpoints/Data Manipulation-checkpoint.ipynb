# Data Manipulation
You may have noticed that often times, the data we retrieve has challenges which need solving before we can engage in data science.
We will discuss here how to manipulate data as sees fit.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# data cleaning as before
df_pop = pd.read_csv('Data/WB_population.csv')
# changing the index column if we didn't do that in the read_csv
df_pop.set_index('Country Name', inplace=True)
# dropping uninteresting columns
df_pop.drop(columns=['Indicator Name', 'Indicator Code', 'Country Code'], inplace=True)
# dropping missing values, default is rows (axis=0) and how='any'
df_pop.dropna(axis=1, how='all', inplace=True)
df_popT = df_pop.T  # or df_pop.transpose()
# set the year index to be a datetime object instead
df_popT.index = pd.to_datetime(df_popT.index, format='%Y')
# a population should not be a float
for col in df_popT.columns:
    df_popT[col] = df_popT[col].astype(int, errors='ignore')
df_popT.head()
```

```python
# sorting data  by column
data = {'Name': ['John', 'Anna', 'Mike', 'Chris'],
        'Age': [28, 24, 35, 45]}
df = pd.DataFrame(data)
sorted_df = df.sort_values(by='Age')
sorted_df
```

```python
# sorting by index
sorted_df = df.sort_index()
sorted_df
```

```python
# by multiple columns
sorted_df = df.sort_values(by=['Age', 'Name'])
sorted_df
```

```python
# filtering
filtered_df = df[df['Age'] > 30]
filtered_df
```

```python
# querying (sql-like syntax)
filtered_df = df.query('Age > 30')
filtered_df
```

```python
# grouping and aggregating
grouped_df = df_popT.groupby(pd.Grouper(freq='10AS')).sum()  # lots of other functions like mean, first, last, etc
grouped_df
```

```python
# alternate approach with aggregates
grouped_df = df_popT.groupby(pd.Grouper(freq='10AS')).agg([np.median, np.sum])  # lots of other functions like mean, first, last, etc
grouped_df
```

```python
# dealing with duplicates
data = {
    'Country': ['USA', 'USA', 'Germany', 'Germany', 'France', 'France'],
    'Year': [2000, 2000, 2001, 2001, 2002, 2002],
    'Population': [300.5, 300.5, np.nan, 82.0, 60.4, np.nan],
    'GDP': [10000, np.nan, 3000, 3000, np.nan, 2500]
}
df = pd.DataFrame(data)
df
```

```python
# marking all bar the first instance as a duplicate
duplicates = df.duplicated(subset='Country')
df[duplicates]
```

```python
df.dropna(inplace=True)  # dropping NaNs
df.drop_duplicates(inplace=True)  # dropping duplicates
df
```

```python
# merging dataframes
data_a = {'ID': [1, 2, 3],
          'Name': ['Alice', 'Bob', 'Charlie']}
data_b = {'ID': [2, 3, 4],
          'Age': [25, 30, 35]}

df_a = pd.DataFrame(data_a)
df_b = pd.DataFrame(data_b)

merged_df = pd.merge(df_a, df_b, on='ID', how='inner')
merged_df
```

```python
merged_df = pd.merge(df_a, df_b, on='ID', how='left')
merged_df
```

```python
# applying
data = {'Name': ['John', 'Anna', 'Mike', 'Chris'],
        'Age': [28, 24, 35, 45]}
df = pd.DataFrame(data)
df['Senior'] = df['Age'].apply(lambda x: 'Yes' if x >= 60 else 'No')
df = df.map(lambda x: float(x) if isinstance(x, int) else x)
df
```

```python
# transforming data
data = {
    'A': [1, 2, 3, 4, 5],
    'B': [10, 20, 30, 40, 50],
    'C': [100, 200, 300, 400, 500]
}
df = pd.DataFrame(data)

def normalise(x):
    return (x - x.mean()) / x.std()

df_normalised = df.transform(normalise)
print(df_normalised)
```

```python
# pivot tables and crosstab
df = pd.DataFrame({
    'A': ['foo', 'foo', 'foo', 'bar', 'bar', 'bar'],
    'B': ['one', 'one', 'two', 'two', 'one', 'one'],
    'C': ['small', 'large', 'large', 'small', 'small', 'large'],
    'D': [1, 2, 2, 3, 3, 4],
    'E': [2, 3, 5, 5, 6, 6]
})
pivot_table_df = pd.pivot_table(df, values='D', index=['A', 'B'], columns=['C'], aggfunc='sum')

# Crosstab Example
crosstab_df = pd.crosstab(df['A'], df['B'])
crosstab_df
```

```python
df.set_index(['A', 'B'], inplace=True)
sliced_data = df.loc[('bar', 'one')]
sliced_data
```

```python
# memory usage and chaining expressions
df.memory_usage(deep=True)
result = (df
          .query('A == "bar"')
          .groupby('B')
          .agg({'D': 'mean', 'E': 'sum'})
          .rename(columns={'D': 'mean_D', 'E': 'sum_E'}))
result
```
