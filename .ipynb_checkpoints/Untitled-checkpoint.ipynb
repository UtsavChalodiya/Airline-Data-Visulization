```python
oyear=pd.read_csv('2023dataset2.csv')
tyear=pd.read_csv('twoyeardata2.csv')
oyear['Date']=pd.to_datetime(oyear['Date'])
tyear['Date']=pd.to_datetime(tyear['Date'])
oyear_grpby_class=oyear.groupby(by=['Date','Class'])['Class'].count().unstack()
oyear_grpby_class['Y']=oyear_grpby_class['Y']*109
oyear_grpby_class['J']=oyear_grpby_class['J']*109
tyear['Y']=tyear['Pax Dom STD']+tyear['Pax Eur STD']+tyear['Pax Int STD']
tyear['J']=tyear['Pax Dom BUS']+tyear['Pax Eur BUS']+tyear['Pax Int BUS']
tyear_group=tyear[['Date','Y','J']].set_index('Date')
result= pd.concat([tyear_group,oyear_grpby_class])
days= list(range(0,1095))
result['Days']=days
result=result.set_index('Days')
x=result.index
y=result['Y']
freq,ccos,csin = scrfft.scrfft(x,y)
freq,ccos,csin=freq[0:7],ccos[0:7],csin[0:7]
y_pred=np.zeros_like(x)
for f,c,s in zip(freq,ccos,csin):
    y_pred+= c*np.cos(2*np.pi*f*x)+s*np.sin(2*np.pi*f*x)
plt.plot(x,y_pred,color='red')
plt.scatter(x,y)
y=result['J']
plt.scatter(x,y)
freq,ccos,csin = scrfft.scrfft(x,y)
freq,ccos,csin=freq[0:7],ccos[0:7],csin[0:7]
y_pred=np.zeros_like(x)
for f,c,s in zip(freq,ccos,csin):
    y_pred+= c*np.cos(2*np.pi*f*x)+s*np.sin(2*np.pi*f*x)
plt.plot(x,y_pred,color='red')
plt.show()
```

```python
def figure4():
    oyear['Weekday']=oyear['Date'].dt.weekday
    tyear['Weekday']=tyear['Date'].dt.weekday
    oyear_d=oyear[oyear['Region']=='Dom']
    oyear_e=oyear[oyear['Region']=='Eur']
    oyear_i=oyear[oyear['Region']=='Int']
    tyear['Dom']=tyear['Pax Dom STD']+tyear['Pax Dom BUS']
    tyear['Eur']=tyear['Pax Eur STD']+tyear['Pax Eur BUS']
    tyear['Int']=tyear['Pax Int STD']+tyear['Pax Int BUS']
    tyear_D=tyear.pivot_table(index='Weekday',values='Dom',columns='Year',aggfunc='sum')
    tyear_E=tyear.pivot_table(index='Weekday',values='Eur',columns='Year',aggfunc='sum')
    tyear_I=tyear.pivot_table(index='Weekday',values='Int',columns='Year',aggfunc='sum')
    oyear_D=oyear_d.pivot_table(index='Weekday',values='Region',columns='Year',aggfunc='count')
    oyear_E=oyear_e.pivot_table(index='Weekday',values='Region',columns='Year',aggfunc='count')
    oyear_I=oyear_i.pivot_table(index='Weekday',values='Region',columns='Year',aggfunc='count')
    Dom=tyear_D.merge((oyear_D)*109,on='Weekday',how='left')
    Eur=tyear_E.merge((oyear_E)*109,on='Weekday',how='left')
    Int=tyear_I.merge((oyear_I)*109,on='Weekday',how='left')
    Dom['Total']=Dom[2021]+Dom[2022]+Dom[2023]
    Eur['Total']=Eur[2021]+Eur[2022]+Eur[2023]
    Int['Total']=Int[2021]+Int[2022]+Int[2023]
    width=0.25
    plt.bar(Dom.index-width,Dom['Total'],width)
    plt.bar(Eur.index,Eur['Total'],width)
    plt.bar(Int.index+width,Int['Total'],width)
```

```python
def figure4():
    # Add weekday
    oyear['Weekday'] = oyear['Date'].dt.weekday
    tyear['Weekday'] = tyear['Date'].dt.weekday

    # Create total passenger columns
    tyear['Dom'] = tyear['Pax Dom STD'] + tyear['Pax Dom BUS']
    tyear['Eur'] = tyear['Pax Eur STD'] + tyear['Pax Eur BUS']
    tyear['Int'] = tyear['Pax Int STD'] + tyear['Pax Int BUS']

    regions = ['Dom', 'Eur', 'Int']
    results = {}

    for r in regions:
        # tyear pivot (sum of passengers)
        t_pivot = tyear.pivot_table(index='Weekday', values=r, columns='Year', aggfunc='sum')

        # oyear pivot (count of flights)
        o_pivot = oyear[oyear['Region'] == r].pivot_table(
            index='Weekday', values='Region', columns='Year', aggfunc='count'
        )

        # Merge and compute total
        df = t_pivot.merge(o_pivot * 109, on='Weekday', how='left')
        df['Total'] = df[2021] + df[2022] + df[2023]

        results[r] = df

    # Plot
    width = 0.25
    plt.bar(results['Dom'].index - width, results['Dom']['Total'], width)
    plt.bar(results['Eur'].index,         results['Eur']['Total'], width)
    plt.bar(results['Int'].index + width, results['Int']['Total'], width)
```

```python
figure4()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

def figure2():
    # Extract time info
    for df in [oyear, tyear]:
        df['Month'] = df['Date'].dt.month
        df['Year'] = df['Date'].dt.year

    # oyear average price
    o_group = oyear.groupby(['Year', 'Month'])['Price'].mean().reset_index(name='Avg_price')

    # tyear average price (including all services)
    tyear['T_price'] = (
        tyear[['Price Dom STD','Price Dom BUS',
               'Price Eur STD','Price Eur BUS',
               'Price Int STD','Price Int BUS']].mean(axis=1)
    )
    t_group = tyear.groupby(['Year', 'Month'])['T_price'].mean().reset_index(name='Avg_price')

    # Combine datasets
    result = pd.concat([o_group, t_group])

    # Month labels
    month_names = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']

    # Plot (single panel)
    plt.figure(figsize=(10, 6))
    sns.barplot(data=result, x='Month', y='Avg_price', hue='Year')

    # Axes formatting
    plt.xticks(ticks=range(12), labels=month_names)
    plt.xlabel('Month')
    plt.ylabel('Average Ticket Price')
    plt.title('Average Ticket Prices Including Additional Services (2021–2023)')

    # Legend
    plt.legend(title='Year')

    # Student ID
    plt.text(0.99, 0.01, 'Student ID: 250',
             transform=plt.gca().transAxes,
             ha='right', va='bottom', fontsize=10)

    plt.tight_layout()
    plt.show()
```

```python
figure2()
```
